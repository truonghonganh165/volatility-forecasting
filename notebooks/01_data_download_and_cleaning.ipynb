```python
import pandas as pd
import yfinance as yf
import numpy as np
import os

print("Libraries loaded successfully.")

# Asset choice
ticker = "^VIX"
safe_ticker = "VIX"

# Download data
data = yf.download(ticker, start="2013-01-01", progress=False)
data = data.dropna()
data.index = pd.to_datetime(data.index)

# Compute log-returns (handle missing Adj Close)
if "Adj Close" in data.columns:
    price_col = "Adj Close"
else:
    price_col = "Close"

print("Using price column:", price_col)

data["log_return"] = np.log(data[price_col]).diff()
data = data.dropna()

# Save cleaned data
raw_path = "../data/raw"
os.makedirs(raw_path, exist_ok=True)

data.to_csv(f"{raw_path}/{safe_ticker}_daily.csv")

print("Saved cleaned data to:", f"{raw_path}/{safe_ticker}_daily.csv")
```
