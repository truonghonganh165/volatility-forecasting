```python
# ===========================================
# Notebook 01 — Data Download & Cleaning
# ===========================================

import pandas as pd
import numpy as np
import yfinance as yf
import os

# Pick asset
ticker = "^VIX"

# Load data
data = yf.download(
    ticker,
    start="2013-01-01",
    progress=False,
    auto_adjust=False
)

# Fix Index
if isinstance(data.columns, pd.MultiIndex):
    data.columns = [col[0] for col in data.columns]
data = data.reset_index()

# Rename Columns 
rename_map = {
    "Date": "Date",
    "Adj Close": "Adj_Close",
    "Close": "Close",
    "Open": "Open",
    "High": "High",
    "Low": "Low",
    "Volume": "Volume"
}
data = data.rename(columns=rename_map)

# Convert Date
data["Date"] = pd.to_datetime(data["Date"])
data = data.sort_values("Date")

# Add log returns
data["log_return"] = np.log(data["Close"]).diff()

# Sanity Checks
print("\nData preview:")
display(data.head())
print("\nData info:")
print(data.info())

# Save final clean CSV
os.makedirs("../data/raw", exist_ok=True)
output_path = "../data/raw/VIX_daily.csv"
data.to_csv(output_path, index=False)
print("\nSaved cleaned data to:", output_path)
```
